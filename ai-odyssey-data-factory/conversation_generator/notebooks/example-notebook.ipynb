## Generar un Dataset de Conversaciones

```python
# Generar dataset con múltiples conversaciones
df = generator.generate_dataset(num_conversations=2)

# Mostrar estadísticas del dataset
print("\nEstadísticas del dataset:")
print(f"Número total de mensajes: {len(df)}")
print(f"Número de conversaciones: {df['conversation_id'].nunique()}")
print("\nDistribución de tipos de conversación:")
print(df['type'].value_counts())
print("\nDistribución de etapas:")
print(df['stage'].value_counts())
```

## Analizar una Conversación Específica

```python
# Seleccionar una conversación específica
conv_id = 0
conversation = df[df['conversation_id'] == conv_id].sort_values('message_id')

print(f"\nConversación {conv_id}:")
for _, row in conversation.iterrows():
    print(f"\n{row['role'].title()} ({row['type']}, {row['stage']}):")
    print(row['message'])
```
